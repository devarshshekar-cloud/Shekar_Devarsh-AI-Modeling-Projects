```python
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline

df = pd.read_json('/content/yelp_academic_dataset_business (1).json', lines = True)

column_drop = ['business_id', 'address', 'postal_code', 'latitude', 'longitude', 'is_open', 'hours', 'attributes', 'categories']
df = df.drop(columns=column_drop)

df.head(10)
```

```python
#Find how many unique states are in this data set:
state_unique = df.state.unique()
state_unique
```

```python
#Find out the average rating in stars for each state with many different businesses in the data set:
df_CA = df[df.state == 'CA']
CAstars_avg = np.mean(df_CA.stars)
print('California Average Rating:' + str(CAstars_avg))

df_MO = df[df.state == 'MO']
MOstars_avg = np.mean(df_MO.stars)
print('Missouri Average Rating:' + str(MOstars_avg))

df_AZ = df[df.state == 'AZ']
AZstars_avg = np.mean(df_AZ.stars)
print('Arizona Average Rating:' + str(AZstars_avg))

df_PA = df[df.state == 'PA']
PAstars_avg = np.mean(df_PA.stars)
print('Pennsylvania Average Rating:' + str(PAstars_avg))

df_TN = df[df.state == 'TN']
TNstars_avg = np.mean(df_TN.stars)
print('Tennessee Average Rating:' + str(TNstars_avg))

df_FL = df[df.state == 'FL']
FLstars_avg = np.mean(df_FL.stars)
print('Florida Average Rating:' + str(FLstars_avg))

df_IN = df[df.state == 'IN']
INstars_avg = np.mean(df_IN.stars)
print('Indiana Average Rating:' + str(INstars_avg))

df_LA = df[df.state == 'LA']
LAstars_avg = np.mean(df_LA.stars)
print('Lousiana Average Rating:' + str(LAstars_avg))

df_NV = df[df.state == 'NV']
NVstars_avg = np.mean(df_NV.stars)
print('Nevada Average Rating:' + str(NVstars_avg))

df_ID = df[df.state == 'ID']
IDstars_avg = np.mean(df_ID.stars)
print('Idaho Average Rating:' + str(IDstars_avg))

df_IL = df[df.state == 'IL']
ILstars_avg = np.mean(df_IL.stars)
print('Illinois Average Rating:' + str(ILstars_avg))

df_NJ = df[df.state == 'NJ']
NJstars_avg = np.mean(df_NJ.stars)
print('New Jersey Average Rating:' + str(NJstars_avg))

df_DE = df[df.state == 'DE']
DEstars_avg = np.mean(df_DE.stars)
print('Delaware Average Rating:' + str(DEstars_avg))
```

```python
rel_states = ['CA', 'FL', 'PA', 'IL', 'NJ', 'AZ', 'TN', 'IN', 'MO', 'LA', 'NV', 'DE']
avg_stars = [(CAstars_avg), (FLstars_avg), (PAstars_avg), (ILstars_avg), (NJstars_avg), (AZstars_avg), (TNstars_avg), (INstars_avg), (MOstars_avg), (LAstars_avg), (NVstars_avg), (DEstars_avg)]


# Create the bar graph
plt.bar(rel_states, avg_stars, color='black')
plt.xlabel('States Ordered by Population')
plt.ylabel('Average Stars')
plt.title('Average Stars vs States Order by Population')
plt.show()

```

```python
#Add in the population of each state:
pop_CA = 39538223
pop_FL = 21538187
pop_PA = 13002700
pop_IL = 12812508
pop_NJ = 9288994
pop_AZ = 7151502
pop_TN = 6910840
pop_IN = 6785528
pop_MO = 6154913
pop_LA = 4657757
pop_NV = 3104614
pop_DE = 989948

state_pop = [pop_CA, pop_FL, pop_PA, pop_IL, pop_NJ, pop_AZ, pop_TN, pop_IN, pop_MO, pop_LA, pop_NV, pop_DE]

#Generate a Scatterplot:
plt.scatter(state_pop, avg_stars)
plt.xlabel('State Population')
plt.ylabel('Average Stars')
plt.title('Average Stars vs State Population')
plt.show()

b,a = np.polyfit(state_pop, avg_stars, deg = 1)
x = np.arange(0,25,0.1)
plt.plot(x, a+b*x, 'r')
plt.show()


```

```python
import statsmodels.api as sm
#Statistical Analysis:

#Correlation
corr_xy = np.corrcoef(state_pop, avg_stars)
print(corr_xy)

#Covariance
cov_xy = np.cov(state_pop, avg_stars)
print(cov_xy)

#Run a Linear Regression
model = sm.OLS(state_pop, avg_stars)
results = model.fit()

print(results.summary())
print(results.params)

#Generate a Scatterplot:
plt.scatter(state_pop, avg_stars)
plt.xlabel('State Population')
plt.ylabel('Average Stars')
plt.title('Average Stars vs State Population')
plt.show()

#Run a Regression Line:
b,a = np.polyfit(state_pop, avg_stars, deg = 1)
x = np.arange(0,25,0.1)
plt.plot(x, a+b*x, 'r')
plt.title('Regression Line')
plt.xlabel('State Population')
plt.ylabel('Average Stars')
plt.show()
```

We can see how there is no correlation between state population and the rating of its business overall.
